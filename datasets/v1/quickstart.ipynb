只读取指标版，不调用 API。不把重复运行当独立任务；缺失的端到端时间保持为空。可在仓库根目录或本 notebook 目录运行。

```python
from pathlib import Path
import importlib.util
folder = Path('datasets/v1') if Path('datasets/v1/use_dataset.py').exists() else Path('.')
spec = importlib.util.spec_from_file_location('release', folder / 'use_dataset.py')
release = importlib.util.module_from_spec(spec)
spec.loader.exec_module(release)
print(release.check(folder))
rows = release.read_rows(folder)
print('Missing end-to-end:', sum(r['end_to_end_ms'] is None for r in rows))
```

```python
for result in release.summaries([r for r in rows if r['batch_id'] == 'scienceworld_v11b']):
    print(result)
pairs = release.paired(rows, 'semantic_top3', 'semantic_flat_retrieval', 'semantic_body_retrieval')
print('Pairs:', len(pairs))
print('Same recorded Skill prompt:', sum(p['same_skill_prompt'] is True for p in pairs))
print('Right minus left total tokens:', sum(p['right_minus_left_total_tokens'] for p in pairs))
```
